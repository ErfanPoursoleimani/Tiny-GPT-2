# Analyse a training run
Set `RUN` to a directory under `outputs/runs/`.

```python
import json, glob
import matplotlib.pyplot as plt
RUN = sorted(glob.glob("../outputs/runs/*"))[-1]
m = json.load(open(RUN + "/metrics.json"))
h = m["history"]
print(RUN); print(json.dumps(m["summary"], indent=1)[:1500])
```

```python
tr = [r for r in h if "train_loss" in r]; va = [r for r in h if "val_loss" in r]
fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
ax[0].plot([r["step"] for r in tr], [r["train_loss"] for r in tr], label="train"); ax[0].plot([r["step"] for r in va], [r["val_loss"] for r in va], "o-", label="val"); ax[0].legend(); ax[0].set_title("loss")
ax[1].plot([r["step"] for r in tr], [r["tokens_per_sec"] for r in tr]); ax[1].set_title("tokens/s"); plt.show()
```
