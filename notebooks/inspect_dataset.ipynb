```python
import sys; sys.path.insert(0, "../src")
from pathlib import Path
import numpy as np
from mini_gpt.data.splits import ensure_tokenized
from mini_gpt.tokenizer.tokenizer import BPETokenizer, tokenizer_dir_for
tok = BPETokenizer.load(tokenizer_dir_for("../data/tokenizer", "wikitext2", 32000))
bins = ensure_tokenized("wikitext2", "../data/processed", tok)
for k, p in bins.items():
    print(k, f"{p.stat().st_size // 2:,} tokens")
```

```python
print(Path("../data/processed/wikitext2/train.txt").read_text()[:1500])
```

```python
arr = np.fromfile(bins["train"], dtype=np.uint16)
print("eos count (documents):", (arr == tok.eos_id).sum())
print(tok.decode(arr[:80].tolist(), skip_special_tokens=False))
```

```python
import collections
top = collections.Counter(arr.tolist()).most_common(15)
print([(tok.decode([i]), n) for i, n in top])
```
