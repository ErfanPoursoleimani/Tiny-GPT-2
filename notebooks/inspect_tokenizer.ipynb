# Inspect the tokenizer

```python
import sys; sys.path.insert(0, "../src")
from mini_gpt.tokenizer.tokenizer import BPETokenizer, tokenizer_dir_for
tok = BPETokenizer.load(tokenizer_dir_for("../data/tokenizer", "wikitext2", 32000))
print(tok.info())
```

```python
for s in ["Artificial intelligence is changing the world.", "Café 日本語 🚀", "unbelievably tokenizable"]:
    ids = tok.encode(s)
    print(len(ids), [tok.decode([i]) for i in ids], tok.decode(ids) == s)
```
